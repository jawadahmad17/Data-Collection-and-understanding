# Business Understanding (Task 10)


High Income Customers: Customers with higher annual income who are more likely to purchase premium products.
Frequent Buyers: Customers who make purchases regularly and contribute significantly to revenue.
Campaign Responders: Customers who accepted previous marketing campaigns.
Customers with Complaints: Customers who submitted complaints and may require better customer service.
Website Visitors: Customers who frequently visit the company's website, showing higher engagement.
High Spending Customers: Customers with large total spending across product categories.
Recently Active Customers: Customers with low recency values, indicating they purchased recently.
